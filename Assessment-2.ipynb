```sql
SELECT category, COUNT(*) FROM workspace.xcart.xcart_sales GROUP BY category HAVING COUNT(*) > 12000;
```

```sql
SELECT category,COUNT(order_item_id) FROM workspace.xcart.xcart_sales WHERE order_status='Cancelled'GROUP BY category ORDER BY COUNT(order_item_id) DESC;
```

```sql
SELECT COUNT(*) FROM (SELECT delivery_city FROM workspace.xcart.xcart_sales GROUP BY delivery_city  HAVING SUM(sales_amount)>60000000);
```

```sql
SELECT payment_mode FROM workspace.xcart.xcart_sales GROUP BY payment_mode HAVING AVG(sales_amount)>6000;
```

```sql
SELECT
COUNT(*),item_label FROM
(SELECT
CASE 
WHEN sales_amount>5000 THEN 'High'
WHEN sales_amount>1000 THEN 'Medium'
ELSE 'Low'
END AS item_label
FROM workspace.xcart.xcart_sales
)
GROUP BY item_label;
```

```sql
SELECT
CASE
WHEN order_day IN ('Saturday','Sunday') THEN 'Weekend'
Else 'Weekday'
END AS day_cat,COUNT(order_item_id), AVG(sales_amount)
FROM workspace.xcart.xcart_sales
GROUP BY day_cat
```

```sql
SELECT 
CASE
WHEN selling_price>=10000 THEN 'Premium'
WHEN selling_price>=1000 THEN 'Mid-Range'
ELSE 'Budget'
END AS price_cat, 
COUNT(*),
SUM(sales_amount)
FROM workspace.xcart.xcart_sales
GROUP BY price_cat

```

```sql
SELECT category,
ROUND(SUM(CASE WHEN order_status = 'Returned' THEN 1 
ELSE 0 END) * 100.0 / COUNT(*)) AS return_rate
FROM workspace.xcart.xcart_sales GROUP BY category
```
